```python
from faker import Faker
import pandas as pd
```

```python
fake=Faker()

users = []
for i in range(100):
    users.append({
        "id": i,
        "email": fake.email(),
        "name": fake.name(),
        "address": fake.address()
    })

users_df = pd.DataFrame(users)
```

```python
users_df.to_csv("users_accounts.csv", index=False)
```
